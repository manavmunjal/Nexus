```python
import pandas as pd
```

```python
df = pd.read_csv('cleaned_data.csv')
```

```python
df.head(10)
```

```python
# randomly change half of "company" column to "BestBuy"
# randomly change half of "product" column to "Appliances"
import random
indices1 = list(range(len(df)))
indices2 = list(range(len(df)))
random.shuffle(indices1)
random.shuffle(indices2)
half = len(df) // 2
for i in indices1[:half]:
    df.at[i, 'company'] = 'BestBuy'
for i in indices2[:half]:
    df.at[i, 'product'] = 'Appliances'
```

```python
df.head(10)
```

```python
df.to_csv('augmented_cleaned_data.csv', index=False, quoting=1)
```
